# GPc tutorials



[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/SheffieldML/notebook/blob/main/GPc/index.ipynb)



Parallel tutorials for the [`gpc`](https://github.com/SheffieldML/GPc) Python façade. The GPy notebooks under [`../GPy/`](../GPy/index.ipynb) remain the **spec/checklist**; these are the GPc ports.



Native C++ demos and markdown in the GPc repo stay first-class ([CIP-0015](https://github.com/SheffieldML/GPc/blob/main/cip/cip0015.md)). This tree is the Jupyter share surface ([CIP-0007](../cip/cip0007.md) in this repo).



Pin: **`gpc>=0.1.0a10`** (TestPyPI until the first PyPI release).



## In subset



* [Basic GP regression](./basic_gp.ipynb) — ↔ `GPy/basic_gp.ipynb` (1-D + 2-D Matérn ARD)

* [Poisson / Student-t (likelihood zoo)](./poisson_regression.ipynb) — ↔ `GPy/Poisson regression tutorial.ipynb` (CIP-003A; `gpc>=0.1.0a10`)

* [Coregionalization / ICM](./coregionalization_toy.ipynb) — ↔ `GPy/coregionalized_regression_tutorial.ipynb`

* [MAP GPLVM (oil_100)](./gplvm_oil100.ipynb) — GPy `gplvm_oil_100`
