# GPc tutorials

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/SheffieldML/notebook/blob/main/GPc/index.ipynb)

Parallel tutorials for the [`gpc`](https://github.com/SheffieldML/GPc) Python façade. The GPy notebooks under [`../GPy/`](../GPy/index.ipynb) remain the **spec/checklist**; these are the GPc ports.

Native C++ demos and markdown in the GPc repo stay first-class ([CIP-0015](https://github.com/SheffieldML/GPc/blob/main/cip/cip0015.md)). This tree is the Jupyter share surface ([CIP-0007](../cip/cip0007.md) in this repo).

## In subset

* [Basic GP regression](./basic_gp.ipynb) — ↔ `GPy/basic_gp.ipynb`
* [Basic kernels / compounds](./basic_kernels.ipynb) — ↔ `GPy/basic_kernels.ipynb`
* [Sparse GP regression](./sparse_gp_regression.ipynb) — ↔ `GPy/sparse_gp_regression.ipynb` (DTCVAR)
* [Basic classification (EP / Laplace)](./basic_classification.ipynb) — ↔ `GPy/basic_classification.ipynb` (`gpc>=0.1.0a6`)
* [Olympic marathon](./olympic_marathon.ipynb) — `pods` lab data (+ sparse DTCVAR beat)
* [MLP vs RBF (olympics)](./mlp_olympic.ipynb) — Williams arcsin prior
* [ExpQuad naming](./expquad_regression.ipynb) — GPy `ExpQuad` paths (CIP-0024)
* [Weight-space ridges](./weightspace_ridge_1d.ipynb) — GPc-specific (CIP-0028)
